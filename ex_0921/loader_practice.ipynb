```python
import base64
import io

from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
from pdf2image import convert_from_path

def image_to_data_url(image) -> str:
    buffer = io.BytesIO()
    image.save(buffer, format = "PNG")
    encoded = base64.b64encode(buffer.getvalue()).decode("utf-8")
    return f"data:image/png;base64,{encoded}"

from pathlib import Path

POPPLER_PATH = str(Path.cwd() / "poppler-26.09.0" / "Library" / "bin")

pages = convert_from_path(
    "data/multimodal practice.pdf",
    dpi=300,
    first_page=1,
    last_page=3,
    poppler_path=POPPLER_PATH,
)

model = ChatOpenAI(model="gpt-4.1-mini", temperature=0.1)
message = HumanMessage(
    content=[
        {
            "type": "text",
            "text": (
                "이 문서 페이지를 Markdown으로 변환하세요."
                "제목과 문단의 계층을 유지하고, 표는 Markdown 표로 작성하며,"
                "차트의 핵심 수치와 추세를 설명하세요."
            ),
        },
        {
            "type": "image_url",
            "image_url": {"url": image_to_data_url(pages[1])},
        },
    ]
)

response = model.invoke([message])
print(response.content)
```

```python
from langchain_core.documents import Document

document = Document(
    page_content=response.content,
    metadata={
        "source": "data/multimodal practice.pdf",
        "page": 1,
        "parser": "multimodal-model",
    },
)

print(document.page_content[:1000])
print(document.metadata)
```
